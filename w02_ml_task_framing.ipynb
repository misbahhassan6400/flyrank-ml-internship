## My Lane as an ML Task

My selected lane is **Refresh / Content Opportunity Scoring**.

This is a **Scoring** problem because the goal is to assign each webpage a score that represents how urgently it should be refreshed. Pages with higher scores are higher priority for updates.


```python
task_type = "Scoring"
print("ML Task Type:", task_type)
```

## Target or Proxy

The target is the **refresh priority score** for each webpage.

If a true refresh label is not available, we can use proxy signals such as:
- Traffic decline
- Ranking decline
- Click-through rate (CTR)
- Content age

These measured signals help estimate which pages are most likely to benefit from a content refresh.

```python
target = "refresh_priority"
print("Target:", target)
```

## Success Metric

A good model should correctly identify pages that need updating.

One suitable metric is **Mean Absolute Error (MAE)** if predicting a score.

If the task is treated as identifying refresh vs. no refresh, metrics such as Precision, Recall, or F1 Score can also be used.

```python
success_metric = "Mean Absolute Error (MAE)"
print("Success Metric:", success_metric)
```

## 5. Why ML Beats a Fixed Rule

A fixed rule, such as refreshing every page older than one year, ignores many important factors. Machine Learning can consider multiple measured signals like search volume, competition, traffic trends, and content performance together. This provides better decision support than using a single if-statement.

```python
reason = "ML combines multiple features instead of relying on one fixed rule."
print(reason)
```
